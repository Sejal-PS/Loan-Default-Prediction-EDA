```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset and read


# df = pd.read_csv(r"D:\Loan Default Prediction Dataset\Loan_default.csv")
df = pd.read_csv(r"/kaggle/input/datasets/nikhil1e9/loan-default/Loan_default.csv")
print("--- Data Shape ---")
print(df.shape)

print("\n--- Data Head ---")
print(df.head())

print("\n--- Column Names ---")
print(df.columns.tolist())


# Data Information
print("\n--- Data Info ---")
df.info()

print("\n--- Data Describe ---")
print(df.describe())



# Default Rate
default_rate = df["Default"].mean() * 100
print(f"\nOverall default rate: {default_rate:.2f}%")



#  Missing Values
print("\n--- Missing Values in Dataset ---")
print(df.isnull().sum())



#  Duplicate Rows
print("\n--- Duplicate Rows ---")
print(f"Total duplicate rows: {df.duplicated().sum()}")
```

```python
# Visualization Settings
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# Default Count
plt.figure()

sns.countplot(
    data=df,
    x="Default",
    hue="Default",
    palette="pastel",
    legend=False
)

plt.title("Loan Default Count")
plt.xlabel("Default (0 = No Default, 1 = Default)")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Loan Amount Distribution
plt.figure()

sns.histplot(
    data=df,
    x="LoanAmount",
    kde=True,
    bins=30
)

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
# Age vs Loan Amount by Default


plt.figure()

sns.scatterplot(
    data=df,
    x="Age",
    y="LoanAmount",
    hue="Default",
    alpha=0.6
)

plt.title("Age vs Loan Amount by Default")
plt.xlabel("Age")
plt.ylabel("Loan Amount")
plt.legend(title="Default")

plt.show()
```

```python
#correlation heatmap


plt.figure(figsize=(12, 8))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```
